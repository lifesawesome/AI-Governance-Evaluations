## 🎯 Objectives

```python
import json
from datetime import datetime

print("📋 Lab 09: Building the AI Evaluation Operating Model")
```

## Step 1: RACI Matrix

Define who is Responsible, Accountable, Consulted, and Informed for each evaluation activity.

```python
raci_matrix = {
    "activities": [
        {
            "activity": "Define evaluation charter",
            "responsible": "AI Engineer",
            "accountable": "Product Owner",
            "consulted": ["Domain Expert", "Risk/Compliance"],
            "informed": ["Engineering Manager"],
            "frequency": "Per project / major model change"
        },
        {
            "activity": "Curate golden dataset",
            "responsible": "AI Engineer",
            "accountable": "AI Engineer",
            "consulted": ["Domain Expert", "QA"],
            "informed": ["Product Owner"],
            "frequency": "Monthly review + ad-hoc additions"
        },
        {
            "activity": "Run pre-release evaluation",
            "responsible": "CI/CD Pipeline (automated)",
            "accountable": "AI Engineer",
            "consulted": [],
            "informed": ["Team (via PR comment)"],
            "frequency": "Every PR / deployment"
        },
        {
            "activity": "Review gate failures",
            "responsible": "AI Engineer",
            "accountable": "Engineering Manager",
            "consulted": ["Domain Expert"],
            "informed": ["Product Owner"],
            "frequency": "Per failure occurrence"
        },
        {
            "activity": "Red-team testing",
            "responsible": "Security Team",
            "accountable": "CISO / Risk Officer",
            "consulted": ["AI Engineer", "Legal"],
            "informed": ["Product Owner", "Engineering Manager"],
            "frequency": "Quarterly + pre-major-release"
        },
        {
            "activity": "Monitor production drift",
            "responsible": "SRE / Platform Team",
            "accountable": "AI Engineer",
            "consulted": [],
            "informed": ["On-call engineer"],
            "frequency": "Continuous (automated)"
        },
        {
            "activity": "Quarterly evaluation review",
            "responsible": "AI Engineer",
            "accountable": "Product Owner",
            "consulted": ["All stakeholders"],
            "informed": ["Leadership"],
            "frequency": "Quarterly"
        },
    ]
}

print("👥 RACI Matrix for AI Evaluation")
print("=" * 80)
print(f"{'Activity':<35} {'Responsible':<20} {'Accountable':<20} {'Frequency'}")
print("-" * 80)
for item in raci_matrix['activities']:
    print(f"{item['activity']:<35} {item['responsible'][:18]:<20} {item['accountable'][:18]:<20} {item['frequency']}")
```

## Step 2: Release Gate Policy

```python
release_policy = {
    "policy_name": "AI Agent Release Gate Policy",
    "version": "1.0",
    "effective_date": datetime.now().strftime("%Y-%m-%d"),
    "environments": {
        "dev": {
            "gates": ["safety_suite"],
            "approval_required": False,
            "auto_deploy": True,
            "description": "Safety checks only for rapid iteration"
        },
        "staging": {
            "gates": ["safety_suite", "quality_suite", "agent_suite"],
            "approval_required": False,
            "auto_deploy": True,
            "min_dataset_coverage": 0.8,
            "description": "Full automated evaluation gate"
        },
        "production": {
            "gates": ["safety_suite", "quality_suite", "agent_suite", "custom_suite"],
            "approval_required": True,
            "approvers": ["AI Engineer", "Product Owner"],
            "auto_deploy": False,
            "min_dataset_coverage": 1.0,
            "regression_check": True,
            "max_regression": 0.05,
            "description": "Full evaluation + human approval + regression check"
        }
    },
    "override_policy": {
        "allowed": True,
        "requires": ["Engineering Manager approval", "Written justification", "Rollback plan"],
        "max_override_duration": "24h",
        "audit_trail": True
    },
    "rollback_triggers": [
        "Safety score > 0 on any metric",
        "Quality regression > 15% within 4 hours of deployment",
        "Customer escalation related to AI response quality"
    ]
}

print("📜 Release Gate Policy")
print("=" * 60)
for env, config in release_policy['environments'].items():
    icon = "🟢" if env == "dev" else "🟡" if env == "staging" else "🔴"
    print(f"\n{icon} {env.upper()}")
    print(f"   Gates: {', '.join(config['gates'])}")
    print(f"   Auto-deploy: {config['auto_deploy']}")
    print(f"   Approval: {'Required' if config['approval_required'] else 'Not required'}")
```

## Step 3: Red-Team Testing Schedule

```python
red_team_schedule = {
    "cadence": "quarterly",
    "additional_triggers": [
        "Before major release (new model, new tools, new domains)",
        "After security incident",
        "When new attack vectors are published"
    ],
    "scope": [
        {
            "category": "Jailbreak attempts",
            "description": "Attempts to override system instructions",
            "test_count": 20,
            "automated": True
        },
        {
            "category": "Data exfiltration",
            "description": "Attempts to extract training data or system prompts",
            "test_count": 15,
            "automated": True
        },
        {
            "category": "Harmful content generation",
            "description": "Attempts to generate unsafe/harmful responses",
            "test_count": 25,
            "automated": True
        },
        {
            "category": "Privilege escalation",
            "description": "Attempts to access unauthorized tools or data",
            "test_count": 10,
            "automated": False
        },
        {
            "category": "Social engineering",
            "description": "Multi-turn manipulation attempts",
            "test_count": 10,
            "automated": False
        },
    ],
    "success_criteria": {
        "max_successful_attacks": 0,
        "reporting_deadline": "5 business days",
        "remediation_deadline": "10 business days for critical"
    }
}

print("🔴 Red-Team Testing Schedule")
print("=" * 60)
print(f"Cadence: {red_team_schedule['cadence']}")
print(f"\nTest Categories:")
total_tests = 0
for scope in red_team_schedule['scope']:
    auto = "🤖" if scope['automated'] else "👤"
    print(f"   {auto} {scope['category']}: {scope['test_count']} tests")
    total_tests += scope['test_count']
print(f"\nTotal tests per cycle: {total_tests}")
```

## Step 4: Governance Dashboard Metrics

```python
governance_dashboard = {
    "name": "AI Evaluation Governance Dashboard",
    "metrics": [
        {
            "metric": "Evaluation Coverage",
            "definition": "% of AI agents with active evaluation pipelines",
            "target": "100%",
            "owner": "Platform Team"
        },
        {
            "metric": "Gate Pass Rate",
            "definition": "% of deployments that pass quality gates on first attempt",
            "target": ">85%",
            "owner": "AI Engineering"
        },
        {
            "metric": "Mean Time to Remediate",
            "definition": "Average time from gate failure to successful re-evaluation",
            "target": "<24 hours",
            "owner": "AI Engineering"
        },
        {
            "metric": "Dataset Freshness",
            "definition": "Days since last golden dataset update",
            "target": "<30 days",
            "owner": "AI Engineer + Domain Expert"
        },
        {
            "metric": "Production Drift Incidents",
            "definition": "Number of drift alerts triggered per month",
            "target": "<3 per month",
            "owner": "SRE"
        },
        {
            "metric": "Red-Team Pass Rate",
            "definition": "% of red-team attacks successfully blocked",
            "target": "100%",
            "owner": "Security"
        },
        {
            "metric": "Override Count",
            "definition": "Number of gate overrides per quarter",
            "target": "<2 per quarter",
            "owner": "Engineering Manager"
        },
    ]
}

print("📊 Governance Dashboard Metrics")
print("=" * 70)
print(f"{'Metric':<30} {'Target':<15} {'Owner'}")
print("-" * 70)
for m in governance_dashboard['metrics']:
    print(f"{m['metric']:<30} {m['target']:<15} {m['owner']}")
```

```python
# Save complete operating model
operating_model = {
    "title": "AI Evaluation Operating Model",
    "version": "1.0",
    "created": datetime.now().strftime("%Y-%m-%dT%H:%M:%SZ"),
    "raci": raci_matrix,
    "release_policy": release_policy,
    "red_team_schedule": red_team_schedule,
    "governance_dashboard": governance_dashboard,
}

with open("./operating-model.json", "w") as f:
    json.dump(operating_model, f, indent=2)

print("\n✅ Complete Operating Model saved to ./operating-model.json")
print("\nThis document serves as the governance foundation for your AI evaluation program.")
```
